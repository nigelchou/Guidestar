### using the MERlin generated bins and computation, assign scores for each spot

```python
num_blanks = 7
num_coding_genes = len(codebook) - num_blanks
```

```python
# load the counts bin
blankHistogram = np.load(os.path.join(destination_folder, 'GenerateAdaptiveThreshold/blank_counts.npy'))
codingHistogram = np.load(os.path.join(destination_folder, 'GenerateAdaptiveThreshold/coding_counts.npy'))
print(blankHistogram.shape == codingHistogram.shape)
print(f"Shape of histogram: {blankHistogram.shape}")
blankFraction = blankHistogram / codingHistogram
blankFraction[codingHistogram == 0] = np.finfo(blankFraction.dtype).max
blankFraction /= num_blanks/(
                num_blanks + num_coding_genes)
blankFraction
```

```python
# recall misid = blankCounts.mean()/codingCounts.mean()
intensity_bins = np.load(os.path.join(destination_folder, 'GenerateAdaptiveThreshold/intensity_bins.npy'))
distance_bins = np.load(os.path.join(destination_folder, 'GenerateAdaptiveThreshold/distance_bins.npy'))
area_bins = np.load(os.path.join(destination_folder, 'GenerateAdaptiveThreshold/area_bins.npy'))

selectData = df[['mean_intensity', 'min_distance', 'area']].to_numpy()
barcodeBins = np.array(
            (np.digitize(selectData[:, 0], intensity_bins,
                         right=True),
             np.digitize(selectData[:, 1], distance_bins,
                         right=True),
             np.digitize(selectData[:, 2], area_bins))) - 1
barcodeBins[0, :] = np.clip(
    barcodeBins[0, :], 0, blankFraction.shape[0]-1)
barcodeBins[1, :] = np.clip(
    barcodeBins[1, :], 0, blankFraction.shape[1]-1)
barcodeBins[2, :] = np.clip(
    barcodeBins[2, :], 0, blankFraction.shape[2]-1)
raveledIndexes = np.ravel_multi_index(
    barcodeBins[:, :], blankFraction.shape)

blankFractionValues = np.take(blankFraction,raveledIndexes)
df['blank_fraction'] = blankFractionValues
```

```python
### need to compute threshold 
from scipy import optimize

def calculate_misidentification_rate_for_threshold(df,blankHistogram, codingHistogram, blankFraction,
            threshold: float) -> float:
        """ Calculate the misidentification rate for a specified blank
        fraction threshold.

        Args:
            threshold: the normalized blank fraction threshold
        Returns: The estimated misidentification rate, estimated as the
            number of blank barcodes per blank barcode divided
            by the number of coding barcodes per coding barcode.
        """
        decodeTask = df
        blankBarcodeCount = 7
        codingBarcodeCount = 133
        blankHistogram = blankHistogram
        codingHistogram = codingHistogram

        selectBins = blankFraction < threshold
        codingCounts = np.sum(codingHistogram[selectBins])
        blankCounts = np.sum(blankHistogram[selectBins])

        return ((blankCounts/blankBarcodeCount) /
                (codingCounts/codingBarcodeCount))
def calculate_threshold_for_misidentification_rate(df,blankHistogram, codingHistogram, blankFraction,targetMisidentificationRate: float) -> float:
        """ Calculate the normalized blank fraction threshold that achieves
        a specified misidentification rate.

        Args:
            targetMisidentificationRate: the target misidentification rate
        Returns: the normalized blank fraction threshold that achieves
            targetMisidentificationRate
        """
        tolerance = 0.001
        def misidentification_rate_error_for_threshold(x, targetError):
            return calculate_misidentification_rate_for_threshold(df,blankHistogram, codingHistogram, blankFraction,x) \
                - targetError
        return optimize.newton(
            misidentification_rate_error_for_threshold, 0.2,
            args=[targetMisidentificationRate], tol=tolerance, x1=0.3,
            disp=False)
```

```python
calculate_threshold_for_misidentification_rate(df,blankHistogram, codingHistogram, blankFraction, 0.05)
```

```python
df['pass_threshold'] = df['blank_fraction'] < calculate_threshold_for_misidentification_rate(df,blankHistogram, codingHistogram, blankFraction, 0.05)
```
