# Stitched DAPI with contrast on LM50GS mouse brain tissue

```python
import tifffile
import numpy as np
import matplotlib.pyplot as plt

import os
# ---- Main folder = the repository folder, which holds Data/ (layout described in Data/README.md). ----
# Run this from the folder the file is in, or set the environment variable GUIDESTAR_MAIN_DIR to the repository path.
main_dir = os.environ.get('GUIDESTAR_MAIN_DIR', '..')


def contrast_stretch(img, low=1, high=99):
    p_low, p_high = np.percentile(img, (low, high))
    return np.clip((img - p_low) / (p_high - p_low), 0, 1)
```

```python
full_dapi = tifffile.imread(os.path.join(main_dir, 'Data/20250728_Lm50GS_mbrain_3cT3/Fused_DAPI.tif'))
```

```python
fig = plt.figure(figsize=(100,100))
plt.imshow(contrast_stretch(full_dapi), cmap='gray')
plt.axis('off')
fig.savefig(os.path.join(main_dir, 'Data/20250728_Lm50GS_mbrain_3cT3/Contrast0_99_Fused_DAPI.svg'), dpi=300)
plt.show()
```
