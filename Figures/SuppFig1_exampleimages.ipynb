```python
def plot_subfig1_images(fov, row_slice, col_slice, plot_scale_bar=True):
    """Set-1 Guidestar genes only (2 x 6 grid). Reconstructed from the saved outputs
    of cells that called it; channel/hyb mapping is identical to plot_subfig1_images_lm50gs."""

    genes = [  # (title, hyb folder, channel index)
        ('hcn1',    'hyb06_1', 0), ('grik3',   'hyb07_1', 0), ('itpr1',   'hyb08_1', 0),
        ('kcna2',   'hyb07_1', 1), ('kcnb1',   'hyb12_1', 2), ('cacna1a', 'hyb13_1', 2),
        ('foxo1',   'hyb06_1', 2), ('scn3a',   'hyb07_1', 2), ('slc1a2',  'hyb10_1', 1),
        ('cdk6',    'hyb12_1', 1), ('cacma1i', 'hyb09_1', 2), ('cacna1e', 'hyb11_1', 2),
    ]

    fig, axes = plt.subplots(2, 6, figsize=(28, 7))

    for ax, (title, hyb, chan) in zip(axes.flat, genes):
        img = tifffile.imread(os.path.join(filepath, f'{hyb}/{hyb}_MMStack_1-Pos_{fov}.ome.tif'))[chan, ...]
        crop = img[row_slice, col_slice]
        im = ax.imshow(crop, cmap='gray', vmin=0, vmax=np.percentile(crop, 99.9))
        ax.set_title(title)

        divider = make_axes_locatable(ax)
        cax = divider.append_axes('right', size='5%', pad=0.05)
        cbar = fig.colorbar(im, cax=cax, orientation='vertical')
        ticks = list(cbar.get_ticks())
        cbar.set_ticks([ticks[0], ticks[-1]])

        ax.set_xticks([])
        ax.set_yticks([])

    if plot_scale_bar:
        scalebar = ScaleBar(122, 'nm', box_color=None, box_alpha=0, color='yellow', location='lower right',
                            width_fraction=0.04, length_fraction=0.3)
        axes[1, 5].add_artist(scalebar)

    fig.tight_layout()
    fig.savefig(f'lm50gsexamples_set1_suppfig1_fov{fov}_{row_slice}_{col_slice}.svg')
    plt.show()
```
